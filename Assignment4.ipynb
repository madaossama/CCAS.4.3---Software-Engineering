````python
prompt_3 = """
Read the following product review and explain how the user's experience
with the product might influence their opinion about recommending it
to others. Include aspects of their satisfaction and any problems they faced.

Product Review:
"The product has a sleek design and functions well for basic tasks.
However, the battery drains quickly, which makes it hard to rely on
during travel. Overall, it's a decent product, but the battery issue is a drawback."
"""
response_3 = generator_1(prompt_3, max_new_tokens=150, do_sample=True, temperature=0.7)
print(response_3[0]['generated_text'])
````

the model assumed is was supposed to write reviews and wrote multiple reviews based on the provided review.

````python
updated_prompt_3 = """
Read the following product review. First summarize the positive and
negative aspects separately. Then explain whether the user would
recommend the product and in which situations. End with one sentence
containing your final recommendation decision.

Product Review:
"The product has a sleek design and functions well for basic tasks.
However, the battery drains quickly, which makes it hard to rely on
during travel. Overall, it's a decent product, but the battery issue is a drawback."
"""
response_4 = response = generator_1(updated_prompt_3, max_new_tokens=150, do_sample=True, temperature=0.7)
print(response_4[0]['generated_text'])
````

after the changes we made to the prompt now its succefully generating a good review and showing both the advantages and disadvantages of the product instead of only showing some advantages with the earlier prompt and sometimes making up things instead of getting its info from the product review attatched
